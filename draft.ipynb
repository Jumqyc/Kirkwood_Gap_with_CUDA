```python
import numpy as np
import matplotlib.pyplot as plt

G = 6.67430e-11  # Gravitational constant in m^3 kg^-1 s^-2

AU = 149597870700 # Astronomical unit in meters
Msun = 1.988475e30
Mjupter = 1898e24  # Mass of Jupiter in kg

print('Mjupter in units of Msun:', Mjupter / Msun)

Msaturn = 568e24  # Mass of Saturn in kg
print('Msaturn in units of Msun:', Msaturn / Msun)

t_day = 24 * 3600  # seconds in a day

print(f'G in units of AU^3 / (Msun * t_day^2): {G/(AU**3 / (Msun * t_day**2))}')
```
